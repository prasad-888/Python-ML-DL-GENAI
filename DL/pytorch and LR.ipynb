# Linear Regression

```python
import torch
import numpy as np
```

```python
inputs = np.array([[73, 67, 43],
                  [91, 88, 64],
                  [87, 134, 58],
                  [102, 43, 37],
                  [69, 96, 70]], dtype='float32')
```

```python
inputs
```

```python
targets = np.array([[56, 81, 119, 22, 103]], dtype='float32')
targets.reshape((-1, 1))
```

```python
targets.shape
```

```python
inputs = torch.from_numpy(inputs)
targets = torch.from_numpy(targets)

inputs[0], targets
```

```python
w = torch.randn(1,3, requires_grad=True)
b = torch.randn(1, requires_grad= True)

w,b
```

```python
def model(x):
    return x @ w.t() + b
```

```python
preds = model(inputs)
preds
```

```python
import torch.nn as nn
```

```python
loss_fn = nn.MSELoss() # Initialize loss function
loss = loss_fn(preds, targets)
loss
```

```python
loss.backward()
w.grad
```

```python
lr = 1e-5
with torch.no_grad():
    w -= w.grad * lr
    b -= b.grad * lr
```

```python
with torch.no_grad():
    w.grad.zero_()
    b.grad.zero_()
```

```python
preds = model(inputs)
loss = loss_fn(preds, targets)
print(loss)
```

```python
def mse(y_pred, y_true):
    error = y_pred - y_true
    return torch.mean(error**2)
```

```python
lr = 1e-5
for i in range(100):
    preds = model(inputs)
    loss = mse(preds, targets)
    loss.backward()
    with torch.no_grad(): # we use no gradient for weight updation
        w -= w.grad * lr
        b -= b.grad * lr
        w.grad.zero_()
        b.grad.zero_()
```

```python
preds = model(inputs)
loss = mse(preds, targets)
print(loss)
```

2️⃣ __init__ — Model construction
def __init__(self, input_dim, output_dim):


input_dim → number of input features

output_dim → number of outputs (targets)

Example:

input_dim = 3
output_dim = 1

🔹 super(Linear_Regression, self).__init__()

This:

Initializes the parent class (nn.Module)

Registers your model properly with PyTorch

Enables:

.parameters()

.to(device)

.train() / .eval()

❌ Without this → model won’t train correctly.

3️⃣ self.linear = torch.nn.Linear(input_dim, output_dim)

This is the entire linear regression model.

Internally, PyTorch creates:

weight: shape (output_dim, input_dim)

bias: shape (output_dim)

Mathematically:

𝑥

And both automatically have:

requires_grad = True

4️⃣ forward(self, x) — Forward pass
def forward(self, x):


This defines how data flows through the model.

PyTorch calls this automatically when you do:

model(x)


You NEVER call forward() directly.

🔹 outputs = self.linear(x)

This applies:

outputs
=
𝑥
𝑊
𝑇
+
𝑏

+b

x: shape (batch_size, input_dim)

outputs: shape (batch_size, output_dim)

Autograd tracks everything here automatically.

🔹 return outputs

Returns predictions.

```python
class Linear_Regression(torch.nn.Module):
    def __init__(self, input_dim, output_dim):
        super(Linear_Regression, self).__init__()
        self.linear = torch.nn.Linear(input_dim, output_dim)
    def forward(self, x):
        outputs = self.linear(x)
        return outputs
```

```python
model = Linear_Regression(3,1)
list(model.parameters())
```

```python
preds = model(inputs)
preds
```

```python
import torch.nn.functional as F

loss_fn = F.mse_loss
loss = loss_fn(model(inputs), targets)
print(loss)
```

```python
opt = torch.optim.SGD(model.parameters(), lr = 1e-5)
```

```python
# utility function to train the model
def fit(num_epochs, model, loss_fn, opt, train_dl):
    for epoch in range(num_epochs):
        for xb,yb in train_dl:
            pred = model(xb)
            loss = loss_fn(pred, yb)
            loss.backward()
            opt.step()
            opt.zero_grad()
        if (epoch+1) % 10 == 0:
            print('Epoch [{}/{}], Loss: {:.4f}'.format(epoch+1, num_epochs, loss.item()))
```

```python
fit(100, model, loss_fn, opt, train_loader)
```

# Gradient Calculation with Autograd

```python
import torch
```

```python
x = torch.randn(3)
print(x)
```

```python
# now we want to calculate gradients of some functions with respect to x we do
x = torch.randn(3, requires_grad=True )
print(x)

y= x+2 
print(y)    # AddBackward0

z = y*y*2
print(z)    # MulBackward

k = y*y*2
k = k.mean()
print(k)     # MeanBackward0

k.backward() # dk/dx
print(x.grad)
```

```python
# now if we dont use requires_grad=True we will get error

c = torch.randn(3, requires_grad=False) # default value is false
print(c)

q = c+2
print(q)

r = q*q*2
# r = r.mean()
print(r)

r.backward()
print(c.grad)
```

```python
c = torch.randn(3, requires_grad= True) # default value is false
print(c)

q = c+2
print(q)

r = q*q*2
# r = r.mean() # scalar value
print(r)


v = torch.tensor([0.1, 1.0, 0.001], dtype=torch.float32) # if no scalar then we have to give it a vector
r.backward(v) #dr/dc
print(c.grad)
```

```python
x = torch.randn(3, requires_grad=True)
print(x)


# WHEN we dont need gradient calculation
# when we have trailing "_" in pytorch then it will modify our variable inplace

# x.requires_grad_(False)
# x.detach()
# with torch.no_grad():
```

```python
x.requires_grad_(False)
print(x)

# we see the requires_grad=True is gone i.e it is False
```

```python
y = x.detach()
print(y)

# here we see y has the same value but doesnt require gradient
```

```python
with torch.no_grad():
    y = x + 2
    print(y)

# so we dont have gradient function output
```

```python
weights = torch.ones(4, requires_grad = True)

for epoch in range(3):
    model_output = (weights*3).sum()

    model_output.backward()

    print(weights.grad)

```

```python
weights = torch.ones(4, requires_grad = True)

for epoch in range(3):
    model_output = (weights*3).sum()

    model_output.backward()

    print(weights.grad)

    weights.grad.zero_()
```

```python
import torch.nn
```

```python
class Linear_Regression(torch.nn.Module):
    def __init__(self, input_dim, output_dim):
        super(Linear_Regression, self).__init__()
        self.linear = torch.nn.Linear(input_dim, output_dim)
    def forward(self, x):
        outputs = self.linear(x)
        return outputs
```

```python
model = Linear_Regression(3,1)
list(model.parameters())
```

```python
preds = model(inputs)
preds
```

```python
import torch.nn.functional as F

loss_fn = F.mse_loss
loss = loss_fn(model(xb).squeeze(dim=1), yb)
print(loss)
```

```python
opt = torch.optim.SGD(model.parameters(), lr=1e-5)
```

```python
def fit(num_epochs, model, loss_fn, opt, train_dl):
    for epoch in range(num_epochs):
        for xb,yb in train_dl:
            pred = model(xb)
            loss= loss_fn(pred.squeeze(dim=1), yb)
            loss.backward() # compute gradient
            opt.step() # update parameters using gradients
            opt.zero_grad() # reset the gradient to zero
        if (epoch+1) % 10 == 0:
            print('Epoch [{}/{}], Loss: {:.4f}'.format(epoch+1, num_epochs, loss.item()))
            
```

```python
fit(100, model, loss_fn, opt, train_loader)
```
