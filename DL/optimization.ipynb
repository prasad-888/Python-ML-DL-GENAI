# Gradient Descent

```python
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt


# Sample Data (for illustration)
X = torch.randn(100, 1)
y = 2*X + torch.randn(100, 1)*0.2`a

#model
model = nn.Linear(1, 1)

# Loss and Optimizer
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)


epochs = 100
losses = []

for epoch in range(epochs):
    optimizer.zero_grad()
    outputs = model(X)
    loss = criterion(outputs, y)
    loss.backward()
    optimizer.step()
    losses.append(loss.item())


plt.plot(losses)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Sgd Loss Curve")
plt.show()
```

# Mini Batch Optimizer

```python
# mini batch gradient descent 
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
from torch.utils.data import DataLoader, TensorDataset

# Sample Data (for illustration)

X = torch.randn(100, 1)
Y = 2*X + torch.randn(100, 1)*0.2

epochs = 100
losses = []

# Model
model = nn.Linear(1,1)

# Loss and Optimzer
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)


dataset = TensorDataset(X, Y)
dataloader = DataLoader(dataset, batch_size=16, shuffle=True)


for epoch in range(epochs):
    for i, data in enumerate(dataloader, 0):
        inputs, targets = data
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, targets)
        loss.backward()
        optimizer.step()
        losses.append(loss.item())


plt.plot(losses)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Mini batch GD Loss Curve")
plt.show()
```

# Mini batch with Momentum


```python
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt

from torch.utils.data import DataLoader, TensorDataset

X = torch.randn(100, 1)
y = 2*X + torch.randn(100, 1)*0.2

epochs = 100
losses = []

# Model
model = nn.Linear(1, 1)

# Loss and Optimizer
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01, momentum=0.9)
dataset = TensorDataset(X, y)
dataloader = DataLoader(dataset, batch_size=16, shuffle = True)

for epoch in range(epochs):
    for i, data in enumerate(dataloader, 0):
        inputs, targets = data
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, targets)
        loss.backward()
        optimizer.step()
        losses.append(loss.item())


plt.plot(losses)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Mini Batch with Momentum Loss Curve")
plt.show()
```

# adagrad optimizer

# RMSprop optimizer

```python
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt

from torch.utils.data import DataLoader, TensorDataset

X = torch.randn(100, 1)
y = 2*X + torch.randn(100, 1)*0.2

epochs = 100
losses = []

# Model
model = nn.Linear(1,1)

# loss and Optimizer
criterion = nn.MSELoss()

optimizer = optim.RMSprop(model.parameters(), lr =0.01, alpha = 0.9)

dataset = TensorDataset(X, y)
dataloader = DataLoader(dataset, batch_size=16, shuffle=True)

for epoch in range(epochs):
    for i, data in enumerate(dataloader, 0):
        inputs, targets = data
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, targets)
        loss.backward()
        optimizer.step()
        losses.append(loss.item())

plt.plot(losses)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("RMSprop loss curve")
plt.show()
```

# Adam optimizer

```python
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt

from torch.utils.data import DataLoader, TensorDataset

X = torch.randn(100, 1)
y = 2*X + torch.randn(100, 1)*0.2

epochs = 100
losses = []

# Model
model = nn.Linear(1,1)

# Loss and Optimizer
criterion = nn.MSELoss()

optimizer = optim.Adam(model.parameters(), lr=0.001)

dataset = TensorDataset(X, y)
dataloader = DataLoader(dataset, batch_size=16, shuffle=True)

for epoch in range(epochs):
    for i, data in enumerate(dataloader, 0):
        inputs, targets = data
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, targets)
        loss.backward()
        optimizer.step()
        losses.append(loss.item())


plt.plot(losses)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Adam Loss Curve")
plt.show()
```
